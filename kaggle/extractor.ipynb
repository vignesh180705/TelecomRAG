````python
MODEL_NAME = "Qwen/Qwen3-14B"

bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.float16,
    bnb_4bit_use_double_quant=True,
)

print("Loading tokenizer...")

tokenizer = AutoTokenizer.from_pretrained(
    MODEL_NAME,
    trust_remote_code=True
)

print("Tokenizer loaded.")
print("Loading Qwen3-14B...")

model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    quantization_config=bnb_config,
    device_map="auto",
    dtype=torch.float16,
    trust_remote_code=True,
)

print("Qwen3-14B loaded!")

print(
    "Allocated:",
    round(torch.cuda.memory_allocated() / 1024**3, 2),
    "GB"
)

print(
    "Reserved:",
    round(torch.cuda.memory_reserved() / 1024**3, 2),
    "GB"
)
````

# Extraction

````python
def extract_one_chunk(chunk):
    inputs=None
    outputs=None
    generated=None
    test_prompt = f"""
Extract technical entities and relationships from TEXT.

RULES:
1. Extract ONLY information explicitly supported by TEXT. Never infer or hallucinate.
2. Use ONLY entity/relationship types defined in SCHEMA; omit uncertain entities.
3. canonical_name = shortest useful name.
4. evidence = exact contiguous quote from TEXT, <=12 words.
5. Relationship source/target MUST exactly match entities in the output.
6. Do not create implicit entities or relationships with missing endpoints.
7. Return empty arrays if nothing valid is found.
8. Return ONLY valid JSON.

SPECIFICATION:
Only formal 3GPP TS/TR explicitly identified in TEXT. Never use for papers,
reports, books, websites, vendors, or bibliography titles.

USE_CASE:
Only when TEXT explicitly identifies a use case, usage/application/service
scenario. Do not infer it from migration, roaming, security, privacy,
resilience, sustainability, energy efficiency, titles, cited works, or questions.

SCHEMA:
{schema_text2}

OUTPUT:
{{
  "entities": [
    {{"canonical_name":"string","entity_type":"allowed type","evidence":"exact quote"}}
  ],
  "relationships": [
    {{"source":"canonical entity","relationship":"allowed relationship",
     "target":"canonical entity","evidence":"exact quote"}}
  ]
}}

TEXT:
{chunk["text"]}
"""

    messages = [
        {
            "role": "system",
            "content": "You are a precise ORAN technical information extraction system. Output JSON only."
        },
        {
            "role": "user",
            "content": test_prompt
        }
    ]

    text = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True,
        enable_thinking=False
    )

    inputs = tokenizer(
        text,
        return_tensors="pt"
    ).to("cuda")

    start = torch.cuda.Event(enable_timing=True)
    end = torch.cuda.Event(enable_timing=True)

    start.record()

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=2048,
            do_sample=False,
            use_cache=True,
            pad_token_id=tokenizer.eos_token_id
        )

    end.record()
    torch.cuda.synchronize()

    elapsed = start.elapsed_time(end) / 1000

    generated = outputs[0][inputs["input_ids"].shape[1]:]

    raw = tokenizer.decode(
        generated,
        skip_special_tokens=True
    ).strip()

    if raw.startswith("```"):
        raw = raw.replace("```json", "", 1)
        raw = raw.replace("```", "", 1).strip()

    try:
        result = json.loads(raw)
    except Exception:
        result = None
    finally:

        del inputs
        del outputs
        del generated

        gc.collect()

        if torch.cuda.is_available():
            torch.cuda.empty_cache()
            torch.cuda.ipc_collect()
    return result, raw, elapsed
````

````python
def extract_many_chunk(chunks):

    result_l = []
    raw_l = []
    elapsed_l = []
    texts = []
    for chunk in chunks:

        test_prompt = f"""
Extract technical entities and relationships from TEXT.

RULES:
1. Extract ONLY information explicitly supported by TEXT. Never infer or hallucinate.
2. Use ONLY entity/relationship types defined in SCHEMA; omit uncertain entities.
3. canonical_name = shortest useful name.
4. evidence = exact contiguous quote from TEXT, <=12 words.
5. Relationship source/target MUST exactly match entities in the output.
6. Do not create implicit entities or relationships with missing endpoints.
7. Return empty arrays if nothing valid is found.
8. Return ONLY valid JSON.

SPECIFICATION:
Only formal 3GPP TS/TR explicitly identified in TEXT. Never use for papers,
reports, books, websites, vendors, or bibliography titles.

USE_CASE:
Only when TEXT explicitly identifies a use case, usage/application/service
scenario. Do not infer it from migration, roaming, security, privacy,
resilience, sustainability, energy efficiency, titles, cited works, or questions.

SCHEMA:
{schema_text2}

OUTPUT:
{{
  "entities": [
    {{"canonical_name":"string","entity_type":"allowed type","evidence":"exact quote"}}
  ],
  "relationships": [
    {{"source":"canonical entity","relationship":"allowed relationship",
     "target":"canonical entity","evidence":"exact quote"}}
  ]
}}

TEXT:
{chunk["text"]}
"""

        messages = [
            {
                "role": "system",
                "content": (
                    "You are a precise ORAN technical information "
                    "extraction system. Output JSON only."
                )
            },
            {
                "role": "user",
                "content": test_prompt
            }
        ]

        text = tokenizer.apply_chat_template(
            messages,
            tokenize=False,
            add_generation_prompt=True,
            enable_thinking=False
        )

        texts.append(text)
    tokenizer.padding_side = "left"

    if tokenizer.pad_token is None:
        tokenizer.pad_token = tokenizer.eos_token
    inputs = tokenizer(
        texts,
        return_tensors="pt",
        padding=True,
        truncation=True
    )
    input_device = model.get_input_embeddings().weight.device

    inputs = {
        k: v.to(input_device)
        for k, v in inputs.items()
    }
    start = torch.cuda.Event(enable_timing=True)
    end = torch.cuda.Event(enable_timing=True)

    start.record()

    with torch.no_grad():

        outputs = model.generate(
            **inputs,
            max_new_tokens=1024,
            do_sample=False,
            use_cache=True,
            pad_token_id=tokenizer.eos_token_id
        )

    end.record()

    torch.cuda.synchronize()

    elapsed_total = start.elapsed_time(end) / 1000
    input_length = inputs["input_ids"].shape[1]
    for i in range(len(chunks)):

        generated = outputs[i][input_length:]

        raw = tokenizer.decode(
            generated,
            skip_special_tokens=True
        ).strip()

        if raw.startswith("```"):
            raw = raw.replace("```json", "", 1)
            raw = raw.replace("```", "", 1).strip()

        try:
            result = json.loads(raw)

        except Exception:
            result = None

        result_l.append(result)
        raw_l.append(raw)

        elapsed_l.append(elapsed_total)
    del inputs
    del outputs

    gc.collect()

    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    return result_l, raw_l, elapsed_l
````
